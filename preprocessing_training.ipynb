```python
import pandas as pd
```

```python
data = pd.read_csv(r'D:\AQI Project\dataset\AQI_ERA5_Ground_Master_AllStations.csv')

data.head()
```

```python
na_data = data[data['latitude'].isna()]["Station"].unique()
print(na_data)
```

```python
data_new = data.drop(columns = ["District" , "Province" , "In_Official_DoE_2025_Report" , "ground_precip_src" , "ground_precip_match_class" , "ground_temp_src" , "ground_temp_match_class" , "has_aqi_pm25" , "has_era5"])

data_new.head()
```

```python
missing_coordinates = {

    'Pulchowk' : (27.682581, 85.318841),
    "TU/Kirtipur" : (27.681790, 85.288250),
    "Dhulikhel" : (27.618330, 85.551940)
}

for station , (lat , long) in missing_coordinates.items():

    data_new.loc[data["Station"] == station , 'latitude'] = lat
    data_new.loc[data["Station"] == station , 'longitude'] = long


data_new = pd.DataFrame(data_new)

data_new.head()
```

```python
print(data_new[data_new["Station"] == "Pulchowk"])
```

```python
lat_lon = data_new[["Station" , "latitude" , "longitude"]].drop_duplicates(subset = "Station")

lat_lon = lat_lon.dropna(subset = ['latitude'])

print(lat_lon.head(20))
```

```python
necessary_stations = ['Bhaktapur' , "Dhulikhel" , 'Pulchowk' , 'Bhaisipati' , 'Ratnapark' , 'Shankhapark' , 'TU/Kirtipur']

df = data_new[data_new['Station'].isin(necessary_stations)]
```

```python
df.to_csv(r'D:\AQI Project\dataset\new_data.csv' , index = False)
```
